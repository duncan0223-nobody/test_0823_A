```python
#以下是將學生分數換算成等級的程式，對比這兩種寫法：
score = int(input("請輸入分數:"))
if score >= 90:
    grade = "優"
else:
    if score >= 80:
        grade = "甲"
    else:
        if score >= 70:
            grade = "乙"
        else:
            if score >= 60:
                grade = "丙"
            else:
                grade = "丁"

print(grade)
```
