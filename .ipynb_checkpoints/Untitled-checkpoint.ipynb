```python
class Person:
    def __init__(self,name:str): #自訂初始化
        self.name = name #定義實體屬性
        
    def introduce(self): #自訂實體方法
        print(f"我的名字是{self.name}")
        
        
```

```python
p1:Person = Person("jenny") #位置呼叫
p2:Person = Person(name="robert") # 名稱呼叫

p1.introduce()
p2.introduce()
```
