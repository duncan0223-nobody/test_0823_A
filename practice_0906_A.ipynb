```python
def calculate_trangle_area(base:float,height:float):
    """
    計算三角型面積.
    參數:
    base(flot):三角型的底邊長度(cm)
    height(float):三角型的高度(cm)
    回傳:
    float:計算出的三角形面積(cm2)
    """
    return (base*height)/2
    
```
